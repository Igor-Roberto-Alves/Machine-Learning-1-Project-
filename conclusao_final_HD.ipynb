#Conclusão Heart Disease
##Análise Exploratória

Este trabalho teve como objetivo analisar diferentes modelos de aprendizado de máquina para prever/ classificar o dataset *Heart Disease*, a fim de preditar se um indivíduo tem tendência a alguma doença cardíaca a partir dos fatores analisados.

Após o tratamento e exploração dos dados do conjunto Heart Disease UCI, a amostra final contém 302 observações válidas, sem valores ausentes e sem duplicações.

A análise exploratória de dados permitiu fazer as seguintes observações:

* Perfil da Amostra:  A amostra tem como perfil pessoas com idade entre 29 e 77 anos, do genero masculino e feminino.A maioria dos pacientes está entre 50 e 60 anos, faixa etária em que as doenças cardiovasculares se tornam mais frequentes. A amostra é composta majoritariamente por homens (≈68%).

* Idade e sexo: Não foi encontrada diferença estatisticamente significativa entre as idades médias de homens e mulheres (p > 0.05), embora as mulheres apresentem leve tendência a serem mais velhas.

* Pressão arterial: A maior parte dos valores se concentra entre 120 e 140 mmHg, sugerindo predominância de pré-hipertensão e hipertensão estágio 1, o que já indica risco aumentado de problemas cardíacos.

* Colesterol: A média de colesterol é de ≈246 mg/dL, valor acima do limite desejável, reforçando o perfil de risco cardiovascular elevado dos pacientes analisados.

* Oldpeak: A maioria dos pacientes apresenta valores próximos de zero, o que indica baixa incidência de isquemia significativa no teste de esforço. No entanto, há poucos casos extremos (outliers) que merecem atenção clínica.

* Número de vasos obstruídos (ca): A distribuição é fortemente assimétrica à direita, com 58% dos pacientes sem obstrução.

* Tipo de dor no peito (cp): O tipo mais frequente é 0 (angina típica), especialmente entre homens. Entre mulheres, observa-se maior diversidade nos tipos de dor.

## Em síntese:
O conjunto de dados representa um perfil típico de pacientes com risco cardiovascular moderado a alto - idade mediana de 55 anos, pressão e colesterol acima dos níveis ideais e predominância masculina. A ausência de associação direta entre número de vasos obstruídos e presença de doença reforça a complexidade multifatorial das cardiopatias, envolvendo fatores fisiológicos, metabólicos e comportamentais.

Essas análises iniciais fornecem insights iniciais para etapas de modelagem.

##Resumo OLS

O modelo de Regressão Linear (OLS) aplicado ao conjunto forneceu uma primeira análise exploratória sobre a relação entre as variáveis clínicas e a presença de doença cardíaca. Apesar de não ser o método mais indicado para uma variável categórica, o ajuste inicial com
𝑅²≈ 0,529, sugerindo um grau razoável de relação linear entre elas.

Após o processo de Backward Selection, que removeu iterativamente os preditores não significativos (p > 0,05), o modelo final manteve desempenho semelhante —
𝑅² = 0.506, acurácia de 0,844 no treino e 0857 no teste.

O modelo reduzido apresenta a seguinte equação:

$\hat y=0,561−0,113(sex)+0,063(thalack)−0,100(oldpeak)−0,100(ca)+0,076(cp1)+0,115(cp2)+0,066(cp3)− 0.059(thal3) - 0,067(slope1)$

As variáveis mais relevantes indicam que ser do sexo masculino, possuir maior oldpeak ou mais vasos principais coloridos (ca) estão associados a uma redução na probabilidade prevista de doença cardíaca, enquanto certos tipos de dor torácica (cp1, cp2, cp3) aumentam essa probabilidade.

De forma geral, o modelo linear simplificado apresentou boa estabilidade, acurácia média de 0,77 na validação cruzada e recall de 0,83, o que demonstra boa capacidade de identificar corretamente os casos positivos. Assim, a regressão linear, embora limitada, serviu como um diagnóstico inicial consistente, orientando a escolha de modelos mais adequados (como a regressão logística) para análises posteriores.

##Resumo LDA e QDA

**O modelo Linear Discriminant Analysis (LDA)** foi aplicado após a etapa de diagnóstico inicial com regressão linear, buscando uma abordagem mais apropriada para a variável target.

Nos dados de teste (30% do total), o modelo apresentou desempenho sólido, com acurácia de 0,791, precisão de 0,788 e sensibilidade (recall) de 0,837, indicando boa capacidade de classificação geral e ótimo desempenho na identificação de pacientes com doença cardíaca.

Durante o treino, o LDA alcançou acurácia de 0,877, o que, comparado ao desempenho de teste, sugere um modelo bem ajustado e sem indícios relevantes de sobreajuste (overfitting).

A validação cruzada 5-fold confirmou a robustez do modelo, apresentando:

Acurácia média: 0,811 ± 0,037

Precisão média: 0,806 ± 0,070

Recall médio: 0,865 ± 0,074

Esses resultados indicam consistência entre os folds, com variações pequenas, o que reforça a estabilidade do LDA. O modelo mostrou excelente equilíbrio entre precisão e sensibilidade, sendo particularmente eficaz na detecção de casos positivos — aspecto essencial em diagnósticos clínicos.

De modo geral, o LDA se mostrou mais adequado que a regressão linear para este problema.

**O modelo Quadratic Discriminant Analysis (QDA)** foi aplicado após o LDA. A flexibilidade torna o QDA mais adaptativo, embora também mais sensível a ruídos e à quantidade de dados disponíveis.

Nos resultados de desempenho:

Treino: acurácia de 0.915, evidenciando ótimo ajuste inicial.

Teste: acurácia de 0.791, precisão de 0.841 e sensibilidade de 0.755, o que demonstra bom equilíbrio entre detecção de casos positivos e controle de falsos positivos, com leve tendência a superajuste em relação ao treino.

Em compensação, o decaimento da acurácia no conjunto de teste demonstra uma chance de overfit.

A validação cruzada 5-fold confirmou a estabilidade do modelo:

Acurácia média: 0.805 ± 0.022

Precisão média: 0.833 ± 0.055

Recall médio: 0.803 ± 0.066


##Resumo Logit

A Regressão Logística foi aplicada como modelo para estimar a probabilidade de ocorrência de doença cardíaca (target = 1) a partir das variáveis clínicas do conjunto heart_ds. Trata-se de um modelo adequado para variáveis categóricas, permitindo interpretar diretamente o efeito de cada preditor sobre as chances da doença.

O modelo inicial, com todas as variáveis, apresentou um Pseudo R² ≈ 0,498, indicando boa capacidade explicativa.

Treino: Acurácia 0,853 | Precisão 0,850 | Recall 0,887 | F1 0,868

Teste: Acurácia 0,879 | Precisão 0,852 | Recall 0,939 | F1 0,893

Esses resultados sugerem alta capacidade de generalização e excelente sensibilidade, com o modelo identificando corretamente a maioria dos casos positivos.

Em seguida, foi aplicado o processo de Backward Selection, removendo iterativamente as variáveis com p-valor > 0,05. O modelo final manteve apenas as variáveis estatisticamente significativas: sex, exang, oldpeak, ca, cp₂, cp₃ e slope₁, resultando em uma estrutura mais simples e estável.

A equação final estimada é:

logit(p)=2,8703−2,1555(sex)−1,2640(exang)−0,8852(oldpeak)−0,8726(ca)+1,5477(cp₂)+1,6479(cp₃)−1,2889(slope₁)

O desempenho manteve-se equilibrado:

Treino: Acurácia 0,829 | Precisão 0,832 | Recall 0,861 | F1 0,846

Teste: Acurácia 0,824 | Precisão 0,789 | Recall 0,918 | F1 0,849

Esses valores confirmam boa generalização e alta sensibilidade, evidenciando a robustez do modelo mesmo após a redução de variáveis. As variáveis com maior impacto negativo nas chances de doença são sex (masculino), exang (angina induzida por exercício), oldpeak e ca, enquanto cp₂ e cp₃ (tipos de dor torácica) aumentam significativamente a probabilidade prevista.

Por fim, aplicou-se a validação cruzada (5-fold) com Backward Elimination dentro de cada fold, garantindo uma avaliação mais robusta da estabilidade do modelo. O processo removeu de forma consistente variáveis como age, fbs, chol, restecg e thal, reforçando que essas não trazem contribuição significativa quando controladas pelas demais.

O desempenho médio obtido foi:

Acurácia: 0.818 ± 0.052 | Precisão: 0.815 ± 0.060 | Recall: 0.866 ± 0.032 | F1: 0.839 ± 0.043

Esses resultados indicam consistência entre os folds, com pequenas variações.

##Resumo Ridge e Lasso

Após o modelo baseline de Regressão Logística, foram aplicadas as versões regularizadas Ridge (L2) e Lasso (L1), com o objetivo de reduzir a multicolinearidade entre as variáveis, controlar o sobreajuste e avaliar a robustez dos coeficientes frente à penalização.

**Ridge Regression (L2)**

O modelo Ridge apresentou excelente desempenho tanto no treino quanto no teste:

Treino: Acurácia 0.886 | Precisão 0.870 | Recall 0.930 | F1 0.899

Teste: Acurácia 0.824 | Precisão 0.851 | Recall 0.816 | F1 0.833

A validação cruzada (5-fold) confirmou a estabilidade dos resultados, com:

Acurácia média: 0.851 ± 0.030 | Precisão: 0.845 ± 0.042 | Recall: 0.890 ± 0.016 | F1: 0.867 ± 0.025

Esses valores indicam excelente equilíbrio entre precisão e sensibilidade, com ligeira queda no teste — o que sugere leve regularização sem perda de generalização.
Os coeficientes mostram efeitos consistentes com o modelo logístico original, mas suavizados pela penalização L2. As variáveis com maior impacto positivo foram cp₂, cp₃ e thal, enquanto sex, exang, ca e oldpeak tiveram forte influência negativa sobre a probabilidade de doença.

Em termos práticos, o Ridge preservou todas as variáveis, apenas reduzindo a magnitude dos coeficientes, o que ajuda a controlar a instabilidade numérica e o risco de sobreajuste.

**Lasso Regression (L1)**

O Lasso apresentou resultados próximos, porém com maior parcimônia ao zerar coeficientes irrelevantes, o que facilita a interpretação e seleção de variáveis.

Treino: Acurácia 0.877 | Precisão 0.862 | Recall 0.922 | F1 0.891
Teste: Acurácia 0.813 | Precisão 0.833 | Recall 0.816 | F1 0.825

Na validação cruzada (5-fold), observou-se desempenho ligeiramente inferior ao Ridge, mas ainda muito consistente:

Acurácia média: 0.841 ± 0.043 | Precisão: 0.829 ± 0.055 | Recall: 0.896 ± 0.038 | F1: 0.860 ± 0.036

Os coeficientes não nulos indicam que o Lasso manteve como preditores mais relevantes:
cp₂, cp₃ e cp₁ (tipos de dor torácica), sex, ca, thal₂, thal₃, thalach e exang, com menor peso para oldpeak e slope₂.
Isso confirma que as variáveis relacionadas à dor torácica e à função cardíaca são as mais discriminantes, enquanto variáveis como age, fbs, restecg e trestbps foram eliminadas, sugerindo menor poder explicativo.

### Comparação Geral dos Modelos

Em síntese, Ridge apresentou o melhor equilíbrio entre recall e estabilidade, enquanto Lasso forneceu maior interpretabilidade ao eliminar preditores redundantes. Ambos confirmam a relevância das variáveis cp₂, cp₃, exang, ca e sex como principais determinantes da presença de doença cardíaca.

Dessa forma, as regressões regularizadas consolidam os achados do modelo logístico, evidenciando que o conjunto de dados é consistente, com baixa variância entre folds, e que a penalização contribui para modelos mais robustos, interpretáveis e generalizáveis.

##Resumo KNN

O modelo KNN foi avaliado com otimização de hiperparâmetros variando o número de vizinhos k e a métrica de distância.

Nos resultados da validação cruzada 5-fold, o modelo apresentou:

Recall médio: 0.843 ± 0.079

F1 médio: 0.839 ± 0.044

A avaliação final com o melhor conjunto de parâmetros confirmou bom equilíbrio entre precisão e sensibilidade, com:

Treino: Acc=0.863 | Prec=0.877 | Rec=0.870 | F1=0.873

Teste: Acc=0.813 | Prec=0.833 | Rec=0.816 | F1=0.825

Esses resultados demonstram que o KNN mantém desempenho competitivo em relação aos modelos lineares (Logístico, Ridge e Lasso), com ligeira perda em generalização, o que é esperado dado seu caráter não paramétrico e sensível à densidade local dos dados.

##Resumo SVM

O modelo SVM com kernel RBF apresentou excelente desempenho no conjunto de treinamento, com alta acurácia (0.924) e F1-Score (0.932), indicando forte capacidade de ajuste aos dados. No conjunto de teste, manteve bons resultados (Acurácia = 0.824; F1 = 0.833), mostrando boa generalização e ausência de overfitting significativo.

A validação cruzada estratificada (5-fold) confirmou essa estabilidade, com médias próximas (Acc = 0.808 ± 0.036; F1 = 0.826 ± 0.036), e variações pequenas entre os folds, o que reforça a consistência do modelo.

##Redes Neurais

A rede neural alcançou acurácia de 0.924 no treino e 0.791 no teste, mostrando bom aprendizado, porém com leve indício de overfitting.
