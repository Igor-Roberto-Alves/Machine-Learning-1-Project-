#Conclusão

Este trabalho teve como objetivo analisar diferentes modelos de aprendizado de máquina para prever/ classificar o dataset *Heart Disease*, a fim de preditar se um indivíduo tem tendência a alguma doença cardíaca a partir de fatores pessoais.

Após o tratamento e exploração dos dados do conjunto Heart Disease UCI, a amostra final contém 302 observações válidas, sem valores ausentes e sem duplicações. O conjunto está bem estruturado, composto por variáveis clínicas e demográficas relevantes para o diagnóstico de doenças cardíacas.

A análise exploratória de dados permitiu fazer as seguintes observações:

* Perfil da Amostra:  A amostra tem como perfil pessoas com idade entre 29 e 77 anos, do genero masculino e feminino.A maioria dos pacientes está entre 50 e 60 anos, faixa etária em que as doenças cardiovasculares se tornam mais frequentes. A amostra é composta majoritariamente por homens (≈68%).

* Idade e sexo: Não foi encontrada diferença estatisticamente significativa entre as idades médias de homens e mulheres (p > 0.05), embora as mulheres apresentem leve tendência a serem mais velhas.

* Pressão arterial: A maior parte dos valores se concentra entre 120 e 140 mmHg, sugerindo predominância de pré-hipertensão e hipertensão estágio 1, o que já indica risco aumentado de problemas cardíacos.

* Colesterol: A média de colesterol é de ≈246 mg/dL, valor acima do limite desejável, reforçando o perfil de risco cardiovascular elevado dos pacientes analisados.

* Oldpeak: A maioria dos pacientes apresenta valores próximos de zero, o que indica baixa incidência de isquemia significativa no teste de esforço. No entanto, há poucos casos extremos (outliers) que merecem atenção clínica.

* Número de vasos obstruídos (ca): A distribuição é fortemente assimétrica à direita, com 58% dos pacientes sem obstrução.

* Tipo de dor no peito (cp): O tipo mais frequente é 0 (angina típica), especialmente entre homens. Entre mulheres, observa-se maior diversidade nos tipos de dor.

## Em síntese:
O conjunto de dados representa um perfil típico de pacientes com risco cardiovascular moderado a alto — idade mediana de 55 anos, pressão e colesterol acima dos níveis ideais e predominância masculina. A ausência de associação direta entre número de vasos obstruídos e presença de doença reforça a complexidade multifatorial das cardiopatias, envolvendo fatores fisiológicos, metabólicos e comportamentais.

Essas análises iniciais fornecem insumos essenciais para etapas posteriores de modelagem preditiva, como a identificação dos principais fatores de risco e a construção de modelos de classificação da presença de doença cardíaca.

# Métodos de Regressão

## Regressão Linear

Após a limpeza dos dados (remoção de duplicatas) obtemos nesta etapa, aplicamos uma regressão linear (OLS) como diagnóstico inicial para avaliar a relação entre as variáveis explicativas do conjunto heart_ds e a variável alvo target, que indica a presença de doença cardíaca.

O modelo apresentou um R² de aproximadamente 0.49, ou seja, cerca de 49% da variação em target é explicada pelas variáveis preditoras. Apesar de a regressão linear não ser o método ideal para variáveis binárias, esse valor fornece uma medida inicial de ajuste razoável.
Em termos de desempenho preditivo, o modelo completo obteve:

* Acurácia (treino): ≈ 0.825
* Acurácia (teste): ≈ 0.791

Em seguida, foi aplicada uma seleção backward com base nos p-valores, removendo iterativamente os preditores com p > 0.05, até restarem apenas as variáveis estatisticamente significativas. Após essa etapa, o novo modelo manteve desempenho semelhante:
 * R² (train) ≈ 0.475

* Acurácia (treino/teste) ≈ 0.825 / 0.769

##Regressão Logistica

Em seguida, a Regressão Logística foi feita e ajustada pelo Backward Selection, apresentando resultados robustos e interpretáveis. O modelo final, com sete preditores relevantes (sex, trestbps, oldpeak, ca, cp2, cp3 e slope1), obteve acurácia média de ≈ 0.82, recall superior a 0.90 e F1-score de ≈ 0.85, destacando-se pela alta sensibilidade — característica essencial em aplicações médicas, onde minimizar falsos negativos é prioritário. A inclusão da regularização (Ridge e Lasso) aumentou a estabilidade e reduziu a complexidade do modelo, com desempenho médio acima de 0.84, consolidando a regressão logística regularizada como uma das abordagens mais eficazes e clinicamente interpretáveis.

O modelo Ridge (L2) apresentou uma acurácia média de 0.851 ± 0.030 e recall médio de 0.890, destacando-se por sua estabilidade e baixa variabilidade entre folds. Esse comportamento indica que a regularização L2 conseguiu reduzir a variância do modelo sem comprometer o ajuste, mantendo todos os coeficientes ativos e distribuindo o peso entre as variáveis correlacionadas. O bom equilíbrio entre precisão (0.845) e recall (0.890) sugere que o Ridge é confiável para detectar corretamente casos positivos sem excessivo número de falsos positivos.

Por outro lado, o modelo Lasso (L1) obteve desempenho praticamente equivalente, com acurácia média de 0.841 ± 0.043 e recall médio de 0.896 na validação cruzada, e acurácia de teste de 0.813. A principal vantagem do Lasso foi a seleção automática de variáveis, eliminando coeficientes irrelevantes (forçando-os a zero) e mantendo apenas as features mais informativas. Entre elas, destacaram-se as relacionadas ao tipo de dor torácica (cp_1, cp_2, cp_3), sexo, e número de vasos principais (ca) — todas clinicamente coerentes com o diagnóstico de doença cardíaca.


#LDA, QDA, KNN, SVM e Redes Neurais

Os modelos LDA e QDA também apresentaram resultados consistentes, com desempenho ligeiramente inferior, confirmando que a separação entre classes é predominantemente linear. A validação cruzada 5-fold reforçou a estabilidade do modelo, com uma acurácia média de 0,791 ± 0,053, precisão média de 0,782 ± 0,067 e recall médio de 0,853 ± 0,069. Esses valores mostram que o LDA generaliza bem. O modelo de Análise Discriminante Quadrática (QDA) obteve acurácia de 0.872 no treino, mas apenas 0.692 no conjunto de teste, indicando possível sobreajuste (overfitting). O resumo final da validação cruzada (QDA) foi:
* Acurácia Média:  0.775 ± 0.050
* Precisão Média:  0.805 ± 0.074
* Recall Médio:    0.765 ± 0.050

O KNN mostrou boa precisão, porém menor recall e maior sensibilidade a ruídos.
Na validação cruzada 5-fold, o modelo alcançou acurácia média de 0.792 ± 0.030, com precisão de 0.806 e recall de 0.817, indicando desempenho consistente entre os folds, mas com variação perceptível no recall. Contudo, ao avaliar no conjunto de teste, houve queda expressiva de desempenho:
* Acurácia: 0.747
* Precisão: 0.842
* Recall: 0.653
* F1: 0.736

Essa redução, especialmente no recall, indica que o KNN teve dificuldade em generalizar para novos dados, classificando corretamente a maioria dos negativos, mas falhando em identificar parte dos positivos (alta taxa de falsos negativos).

O modelo SVM apresentou acurácia de 0.708 no conjunto de teste, desempenho inferior aos modelos lineares regularizados (Ridge e Lasso) e também abaixo do KNN.

A rede neural atingiu acurácia de 0.844 no treino e 0.824 no teste, com early stopping na epoch 62, o que indica boa generalização e evitou overfitting. O decaimento da loss de 0.6775 para 0.5513 mostra uma convergência estável, e a parada precoce confirma que o modelo aprendeu o padrão subjacente sem memorizar o conjunto de treino.


Com base nos resultados apresentados, os modelos de regressão linear e logística fornecem uma boa visão inicial, mas a regressão logística regularizada (Ridge ou Lasso) se destaca como a abordagem mais adequada para este problema. Ela combina alta acurácia, excelente recall (essencial para minimizar falsos negativos em contexto médico) e interpretabilidade clínica, além de estabilidade e baixa variância graças à regularização. Enquanto métodos como QDA, KNN, SVM e redes neurais apresentaram limitações — seja por sobreajuste, baixa generalização ou menor sensibilidade a casos positivos —, a regressão logística regularizada equilibra desempenho preditivo robusto, simplicidade e relevância clínica das variáveis, tornando-se a escolha ideal para modelar a probabilidade de ocorrência de doenças cardíacas no conjunto de dados analisado.
