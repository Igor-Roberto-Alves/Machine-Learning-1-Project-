#Relatório Flights

Resumo:

No nosso Dataset temos informações sobre vôos entre diversas cidades da Índia, sendo os preditores: a companhia aérea, o trajeto entre as cidades de origem e destino, a classe escolhida, o número de escalas, a duração do voo e a antecedência com que a passagem é comprada. O desafio é entender como cada um desses aspectos pesa na hora de definir quanto o passageiro vai pagar no final.

Nosso objetivo é conseguir prever o preço da passagem a partir desses preditores, e entender quais aspectos tem mais peso nessa decisão


##Primeiros insights sobre os preditores



```python
import pandas as pd

# Criando o DataFrame para o modelo Flights
flights_data = {
    'Preditor': [
        'airline', 'flight', 'source_city', 'departure_time', 'stops',
        'arrival_time', 'destination_city', 'class', 'duration', 'days_left', 'price'
    ],
    'Descrição': [
        'Nome da companhia aérea responsável pelo voo.',
        'Código ou número identificador do voo.',
        'Cidade de origem de onde o voo parte.',
        'Faixa de horário da decolagem',
        'Número de paradas intermediárias (0, 1, 2).',
        'Faixa de horário de chegada ao destino.',
        'Cidade de destino para onde o voo está indo.',
        'Classe da passagem (Economy, Business).',
        'Duração total do voo em horas.',
        'Quantidade de dias restantes até a data do voo no momento da compra',
        'Preço da passagem aérea.'
    ],

    'Tipo': [
        'Nominal', 'Nominal', 'Nominal', 'Nominal', 'Numérico',
        'Nominal', 'Nominal', 'Nominal', 'Numérico', 'Numérico', 'Numérico'
    ]
}

# Criando o DataFrame
df_flights = pd.DataFrame(flights_data)

# Exibindo o DataFrame
df_flights

```

Pela análise exploratória dos dados, podemos inferir algumas tendencias:

* A companhia aerea mais usada é a Vistara
* A cidade de origem mais comum é Dehli e Mumbais, mas elas tem uma distribuição relativamente próxima.
* Existe uma tendencia maior a saide de manhã ou manhã cedo
* A maioria dos voos tem uma parada
* Existe uma tendencia maior a chegada durante a tarde/noite
* As cidades de destino mais comum são Mumbai e Dehli, mantendo ainda assim uma distribuição proxima com as demais
* A grande maioria das passagens é na classe economica
* A distribuição de antecedencia na compra da passagem é quase igual para todos os intervalos, com ligeira diminuição para compras com menos de um dia de antecedencia
* A duração do voo se concentra em torno de 10 horas
* O preço tem uma mediana com cerca de 7500 reais, e se concentra em valores mais baixos, mas tem alta dispersão (passagens consideravelmente mais caras)

------------------------------------------------------------------

A duração parece ter influencia alta no preço, pelo mapa de calor feito.

Vistara é a companhia com a maior mediana de preços e também com a maior dispersão (amplitude interquartil). AirAsia e SpiceJet apresentam as menores medianas, confirmando a percepção de que são companhias de custo mais baixo no dataset.

O horário de chegada e saída também parece ter relação forte com o preço, e rotas chegando tarde da noite são mais baratas normalmente.

Os mapas não demonstram correlação direta entre distância física e preço. Isso indica a ideia de que a demanda do mercado, a competição entre companhias e a qualidade do serviço (classe e número de escalas) são preditores mais importantes do que a distância.

A analise também mostra forte relação da escolha da companhia aerea para influencia do preço. Porém essa escolha tem forte correlação com outros preditores (linhas aereas maiores costumam vender mais, tem mais rotas, fazem voos mais longos e possuem aviões com maior capacidade para variação de classes, enquanto linhas aereas menores e de baixo custo costumam ter pouca emissão de passagens executivas, fazem trajetos menores, em horarios mais espaçados e com menos oferta)
